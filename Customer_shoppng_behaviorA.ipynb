```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))

```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns = {'purchase_amount_(usd)' : 'purchase_amount'})
```

```python
df.isnull().sum()
```

```python
df.columns
```

```python
#create a column age group 
labels = ['Young Adult', 'Adult', ' Middle-Aged', 'Senior']
df['Age_group'] = pd.qcut(df['age'], q=4 , labels = labels)
```

```python
df[['age','Age_group']].head(10)
```

```python
df.columns
```

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Bi-Weekly': 14,
    'Monthly': 30,
    'Quarterly': 90,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days', 'frequency_of_purchases']].head(10)
```

```python
import pandas as pd
from sqlalchemy import create_engine
from urllib.parse import quote_plus

# ----------------------------
# MySQL Connection Details
# ----------------------------
username = "root"
password = quote_plus("Anmol@123")   # Encode special characters like @
host = "localhost"
port = 3306
database = "mydatabase"

# ----------------------------
# Create Engine
# ----------------------------
engine = create_engine(
    f"mysql+pymysql://{username}:{password}@{host}:{port}/{database}"
)

# ----------------------------
# Test Connection
# ----------------------------
try:
    with engine.connect() as conn:
        print("✅ Connected to MySQL successfully!")

    # ----------------------------
    # Write DataFrame to MySQL
    # ----------------------------
    table_name = "mytable"

    df.to_sql(
        name=table_name,
        con=engine,
        if_exists="replace",   # replace existing table
        index=False            # don't write DataFrame index
    )

    print("✅ DataFrame uploaded successfully!")

    # ----------------------------
    # Read Data Back
    # ----------------------------
    result = pd.read_sql(f"SELECT * FROM {table_name} LIMIT 5;", engine)

    print(result)

except Exception as e:
    print("❌ Error:")
    print(e)
```
